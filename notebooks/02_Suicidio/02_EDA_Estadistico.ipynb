## DETECCIÓN DE VALORES ATÍPICOS - MÉTODO IQR

```python
# DETECCIÓN DE VALORES ATÍPICOS - MÉTODO IQR

q1 = df_both["suicide_rate_per_100k"].quantile(0.25)
q3 = df_both["suicide_rate_per_100k"].quantile(0.75)

iqr = q3 - q1

limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

outliers = df_both[
    (df_both["suicide_rate_per_100k"] < limite_inferior) |
    (df_both["suicide_rate_per_100k"] > limite_superior)
]

print("Q1:", round(q1, 2))
print("Q3:", round(q3, 2))
print("IQR:", round(iqr, 2))
print("Límite inferior:", round(limite_inferior, 2))
print("Límite superior:", round(limite_superior, 2))
print("Cantidad de valores atípicos:", len(outliers))
print(
    "Porcentaje de valores atípicos:",
    round(len(outliers) / len(df_both) * 100, 2),
    "%"
)
```

## Valores atípicos mediante el método IQR

Para identificar posibles valores atípicos en la tasa de suicidio se utiliza el método del rango intercuartílico (IQR). Este criterio considera como potencialmente atípicos los valores que se encuentran por debajo de Q1 \- 1.5 × IQR o por encima de Q3 \+ 1.5 × IQR.

El análisis se realiza sobre los registros consolidados (sex = both) para evitar representar varias veces una misma combinación país\-año.

### Detección de valores atípicos mediante el método IQR

Para identificar posibles valores atípicos en la tasa de suicidio se utiliza el método del rango intercuartílico (IQR). Este método define como potencialmente atípicas aquellas observaciones inferiores a `Q1 - 1.5 × IQR` o superiores a `Q3 + 1.5 × IQR`.

El análisis se realiza sobre los registros consolidados (`sex = both`) para evitar representar varias veces una misma combinación país-año. La presencia de un valor atípico no implica necesariamente un error en los datos; puede representar una observación extrema que requiere atención durante el análisis.

```python
# ============================================================
# DETECCIÓN DE VALORES ATÍPICOS - MÉTODO IQR
# ============================================================

variable = "suicide_rate_per_100k"

q1 = df_both[variable].quantile(0.25)
q3 = df_both[variable].quantile(0.75)

iqr = q3 - q1

limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

outliers = df_both[
    (df_both[variable] < limite_inferior) |
    (df_both[variable] > limite_superior)
]

porcentaje_outliers = len(outliers) / len(df_both) * 100

print(f"Q1: {q1:.2f}")
print(f"Q3: {q3:.2f}")
print(f"IQR: {iqr:.2f}")
print(f"Límite inferior: {limite_inferior:.2f}")
print(f"Límite superior: {limite_superior:.2f}")
print(f"Cantidad de valores atípicos: {len(outliers)}")
print(f"Porcentaje de valores atípicos: {porcentaje_outliers:.2f}%")
```

## Interpretación de asimetría y curtosis

La variable suicide\_rate\_per\_100k presenta una **asimetría positiva de 3.152**, lo que indica que su distribución está marcadamente sesgada hacia la derecha. Esto sugiere que la mayoría de las observaciones se concentra en tasas relativamente bajas o intermedias, mientras que un número menor de registros presenta tasas considerablemente más elevadas.

La **curtosis de 16.348** es considerablemente superior a cero, lo que indica una distribución leptocúrtica, con colas pesadas y una presencia importante de valores extremos. Este resultado es consistente con los valores atípicos identificados previamente mediante el método IQR.

En conjunto, ambos indicadores muestran que la variable no presenta una distribución simétrica y que existen observaciones extremas que deben considerarse en el análisis. Estos resultados serán examinados posteriormente mediante histogramas y diagramas de caja en el EDA visual.

El coeficiente de correlación permite medir la dirección y la intensidad de una relación lineal. Un valor cercano a 1 representa una asociación positiva fuerte, un valor cercano a \-1 representa una asociación negativa fuerte y un valor próximo a 0 representa una relación lineal débil.

La existencia de una correlación no implica causalidad. Por lo tanto, los resultados obtenidos se utilizan para identificar asociaciones que posteriormente serán examinadas mediante gráficos de dispersión y una matriz de calor.
